```python
import numpy as np

prices = np.array([19.99, 45.50, 3.25, 89.00, 12.75, 67.30])
quantities = np.array([2, 1, 10, 1, 4, 3])

# (a) Calculate the total cost for each item (price × quantity) — one line, no loop
total_cost = prices * quantities
print(total_cost)

# (b) Apply a 10% discount to all prices and store in a new array
discount_prices = prices * 0.9
print(discount_prices)

# (c) Using comparison, create a boolean array showing which items cost more than $20
boolean_prices = prices > 20
print(boolean_prices)

# (d) Calculate the total revenue (sum of all price × quantity) in one line
total_revenue = (prices.sum() * quantities).sum()
print(total_revenue)

# (e) Find the average price and the most expensive item's price using built-in aggregate methods
print(prices.mean())
print(prices.max())
```
